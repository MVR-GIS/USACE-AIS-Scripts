```python
import requests
import pandas as pd

# 1. Define the Service URL (Append the layer ID, e.g., /0 for points)
# Based on your email, try layer 0 or 1
base_url = "https://emvrw07v05a0a05.mvr.ds.usace.army.mil/b5arcgis/rest/services/Transportation/MVR_Dredging_Layers/FeatureServer/0/query"

# 2. Define parameters for the request
params = {
    'where': '1=1',           # Get all records
    'outFields': '*',         # Get all columns (including placementAreaIDPK)
    'f': 'json',              # Return format
    'returnGeometry': 'true', # Set to true if you want Lat/Lon
    'outSR': '4326'           # Return coordinates in WGS84 (Lat/Lon)
}

try:
    print("Connecting to MVR EGIS Server...")
    response = requests.get(base_url, params=params, verify=False) # verify=False if SSL certs are internal
    data = response.json()

    # 3. Parse the features into a list
    features = data.get('features', [])
    
    if not features:
        print("No data found or check the Layer ID.")
    else:
        # Flatten the JSON (extract attributes and geometry)
        rows = []
        for f in features:
            attr = f['attributes']
            if 'geometry' in f:
                # If it's a point, add X and Y
                attr['longitude'] = f['geometry'].get('x')
                attr['latitude'] = f['geometry'].get('y')
            rows.append(attr)

        # 4. Create DataFrame and Export
        df = pd.DataFrame(rows)
        output_file = "test_DMMP_Sites_Export.csv"
        df.to_csv(output_file, index=False)
        print(f"Success! Exported {len(df)} records to {output_file}")

except Exception as e:
    print(f"An error occurred: {e}")
```
